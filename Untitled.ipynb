```python
import re
import math
import numpy as np
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import sent_tokenize, word_tokenize
from sklearn.feature_extraction.text import TfidfVectorizer
import networkx as nx
from transformers import pipeline
from PyPDF2 import PdfReader
import gradio as gr
```

```python
nltk.download('punkt')
nltk.download('stopwords')
```

```python
STOPWORDS = set(stopwords.words("english"))

def clean_text(text):
    text = text.replace("\n", " ")
    text = re.sub(r'\s+', ' ', text)
    return text.strip()

def normalize_sentence(sentence):
    sentence = sentence.lower()
    sentence = re.sub(r'[^a-z0-9\s]', '', sentence)
    tokens = [t for t in word_tokenize(sentence) if t not in STOPWORDS]
    return " ".join(tokens)

def extract_text_from_pdf(pdf_path):
    """Extract all text from a PDF file"""
    text = ""
    reader = PdfReader(pdf_path)
    for page in reader.pages:
        text += page.extract_text() + " "
    return clean_text(text)
```

```python
def build_similarity_matrix(sentences):
    normalized = [normalize_sentence(s) for s in sentences]
    vectorizer = TfidfVectorizer()
    tfidf = vectorizer.fit_transform(normalized)
    sim_matrix = (tfidf * tfidf.T).toarray()
    np.fill_diagonal(sim_matrix, 0)
    return sim_matrix

def textrank_summary(text, summary_ratio=0.2):
    sentences = sent_tokenize(clean_text(text))
    if len(sentences) < 2:
        return text
    
    sim_matrix = build_similarity_matrix(sentences)
    nx_graph = nx.from_numpy_array(sim_matrix)
    scores = nx.pagerank(nx_graph)
    ranked = sorted(((scores[i], s) for i, s in enumerate(sentences)), reverse=True)
    
    top_n = max(1, int(len(sentences) * summary_ratio))
    top_sentences = [sent for _, sent in ranked[:top_n]]
    return " ".join(top_sentences)
```

```python
summarizer = pipeline("summarization", model="sshleifer/distilbart-cnn-12-6")

def abstractive_summary(text, min_length=30, max_length=120):
    text = clean_text(text)
    if len(text) > 2000:
        text = text[:2000]  # shorten overly long inputs
    result = summarizer(text, max_length=max_length, min_length=min_length, do_sample=False)
    return result[0]['summary_text']
```

```python
def summarise_pdf(pdf_file, method):
    text = extract_text_from_pdf(pdf_file.name)
    if len(text) < 50:
        return "PDF is empty or not readable."
    
    if method == "Extractive":
        return textrank_summary(text, summary_ratio=0.3)
    else:
        return abstractive_summary(text)
```

```python
demo = gr.Interface(
    fn=summarise_pdf,
    inputs=[
        gr.File(label="Upload Your Notes (PDF only)", file_types=[".pdf"]),
        gr.Radio(["Extractive", "Abstractive"], label="Choose Summarization Method")
    ],
    outputs="text",
    title="🧠 Notes Summariser (PDF Upload)",
    description="Upload your PDF notes and get a clean, AI-generated summary using NLP."
)

demo.launch()
```
